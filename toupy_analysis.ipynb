```python
ddw_images = np.load(r"data/ddw_data.npy")
icecream_images = np.load(r"data/icecream_data.npy")
gt_images = np.load(r"data/gt_data.npy")
fpb_images = np.load(r"data/fpb_data.npy")
```

```python
ice_FSC = FSCPlot(icecream_images[0], gt_images[0], pixel_size=10)
fpb_FSC = FSCPlot(fpb_images[0], gt_images[0], pixel_size=10)
ddw_FSC = FSCPlot(ddw_images[0], gt_images[0], pixel_size=10)
```

```python
for i, angle in enumerate([51.0, 48.0, 0, -47.99, -50.99]):
    ice_FSC = FSCPlot(icecream_images[i], gt_images[i], pixel_size=10)
    # fpb_FSC = FSCPlot(fpb_images[i], gt_images[i], pixel_size=10)
    ddw_FSC = FSCPlot(ddw_images[i], gt_images[i], pixel_size=10)

    ice_FSC_scores = ice_FSC.FSC
    ddw_FSC_scores = ddw_FSC.FSC

    voxel_size = 10.0
    box_size_px = len(ice_FSC_scores) * 2
    spatial_frequencies = np.arange(1, len(ice_FSC_scores) + 1) / (box_size_px * voxel_size)
    plt.plot(spatial_frequencies, ice_FSC_scores, label="Icecream")
    plt.plot(spatial_frequencies, ddw_FSC_scores, label="DDW")
    plt.legend()
    plt.xlabel("Spatial frequency (1/nm)")
    plt.ylabel("FSC")
    plt.grid(True, alpha=0.3)
    plt.title(f"FSC Comparison at {angle} degrees")
    plt.show()
```

```python


for i, angle in enumerate([51.0, 48.0, 0, -47.99, -50.99]):
    if i == 2:
        continue
    ice_FSC = FSCPlot(icecream_images[i], gt_images[i], pixel_size=10)
    # fpb_FSC = FSCPlot(fpb_images[i], gt_images[i], pixel_size=10)
    ddw_FSC = FSCPlot(ddw_images[i], gt_images[i], pixel_size=10)

    ice_FSC_scores = ice_FSC.FSC
    ddw_FSC_scores = ddw_FSC.FSC

    if i == 0:
        ddw_average_FSC = ddw_FSC_scores
        ice_average_FSC = ice_FSC_scores
    else:
        ddw_average_FSC += ddw_FSC_scores
        ice_average_FSC += ice_FSC_scores

ddw_average_FSC /= len([51.0, 48.0, 0, -47.99, -50.99])-1
ice_average_FSC /= len([51.0, 48.0, 0, -47.99, -50.99])-1

voxel_size = 10.0
box_size_px = len(ice_FSC_scores) * 2
spatial_frequencies = np.arange(1, len(ice_FSC_scores) + 1) / (box_size_px * voxel_size)


plt.plot(spatial_frequencies, ice_average_FSC, label="Icecream")
plt.plot(spatial_frequencies, ddw_average_FSC, label="DDW")
plt.legend()
plt.xlabel("Spatial frequency (1/nm)")
plt.ylabel("FSC")
plt.grid(True, alpha=0.3)
plt.title(f"Average FSC Comparison at excluded tilt angles")
plt.show()
```

```python
voxel_size = 10.0
box_size_px = len(ice_FSC_scores) * 2
r = np.arange(1, len(ice_FSC_scores) + 1)
spatial_frequencies = r / (box_size_px * voxel_size)

resolution_max = 1000.0  # Å; set to None to plot the full range up to Nyquist

if resolution_max is not None:
    keep = spatial_frequencies >= 1 / resolution_max
else:
    keep = np.ones_like(spatial_frequencies, dtype=bool)

freqs = spatial_frequencies[keep]

plt.plot(freqs, ice_average_FSC[keep], label="Icecream")
plt.plot(freqs, ddw_average_FSC[keep], label="DDW")
plt.title(f"Average Fourier Shell Correlation (FRC) on left out angles",
          fontsize=14, fontweight="bold")
plt.xlabel("Spatial Frequency ($Å^{-1}$)", fontsize=12)
plt.ylabel("FSC", fontsize=12)
plt.ylim(-1.1, 1.1)
plt.grid(True, linestyle=":", alpha=0.6)

ax1 = plt.gca()
ax2 = ax1.twiny()
positive_frequencies = freqs[freqs > 0]
tick_freqs = np.linspace(positive_frequencies[0], positive_frequencies[-1], 5)
tick_resolutions = [f"{1 / f:.1f} Å" for f in tick_freqs]

ax2.set_xlim(ax1.get_xlim())
ax2.set_xticks(tick_freqs)
ax2.set_xticklabels(tick_resolutions)
ax2.set_xlabel("Resolution (Å)", fontsize=12)

ax1.legend(loc="lower left")
plt.tight_layout()
plt.show()
```
